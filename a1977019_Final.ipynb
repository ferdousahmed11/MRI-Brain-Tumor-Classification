```python
# # Brain Tumor MRI Classification - Part 1: Preprocessing and Baseline Models
#
# This notebook covers:
# 1. Dataset loading and preprocessing
# 2. Exploratory Data Analysis (EDA)
# 3. Training three pretrained baseline models: VGG16, ResNet50, DenseNet121
# 4. Evaluation and saving results to a JSON file
#
# **Dataset:** Brain Tumor MRI Dataset (Kaggle) - 4 classes: glioma, meningioma, pituitary, notumor
#
# **How to run:**
# - Place the `dataset` folder in the same directory as this notebook.
# - Run all cells from top to bottom.
# - Results are saved to `results/baseline_results.json` for use in Part 2.


# ## 1. Imports and Configuration


import os
import json
import random
import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
from torchvision import datasets, transforms, models
from PIL import Image

from sklearn.model_selection import train_test_split
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, confusion_matrix, classification_report,
                             roc_auc_score)

print("PyTorch version:", torch.__version__)
print("GPU available:", torch.cuda.is_available())


# ---------------- Configuration ----------------
# All settings are kept in one place so they are easy to find and change.

DATA_DIR    = "/content/dataset/dataset/classification_task"   # contains train/ and test/
RESULTS_DIR = "/content/results"   # folder where JSON results are saved

IMG_SIZE   = 224             # input image size for the models
BATCH_SIZE = 32
EPOCHS     = 15
LR         = 1e-4            # learning rate for fine-tuning pretrained models
SEED       = 42              # fixed seed so results are reproducible

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
os.makedirs(RESULTS_DIR, exist_ok=True)

print("Using device:", DEVICE)


# Set the random seed everywhere so both notebooks use the same data split.
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)


# ## 2. Dataset Loading and Preprocessing
#
# The Kaggle dataset comes with `Training` and `Testing` folders. We use the `Training`
# folder and split it into training and validation sets, and keep `Testing` as the
# untouched test set.


# Print the folder structure so the dataset layout is easy to confirm.
for root, dirs, files in os.walk(DATA_DIR):
    depth = root.replace(DATA_DIR, "").count(os.sep)
    if depth <= 1:
        print("  " * depth + os.path.basename(root) + f"/  ({len(files)} files)")


# ---------------- Preprocessing / transforms ----------------
# Training transform includes augmentation to reduce overfitting.
# Validation and test transforms only resize and normalise (no augmentation).

MEAN = [0.485, 0.456, 0.406]   # ImageNet normalisation values
STD  = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])

eval_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])


# ---------------- Load the dataset ----------------
# Some copies of this dataset have Training/Testing folders, others have the four
# class folders directly inside `dataset`. Both layouts are handled here.

TRAIN_NAMES = ["train", "training"]
TEST_NAMES  = ["test", "testing"]


def _match_subdir(directory, wanted_names):
    """Find a subfolder whose name matches one of wanted_names (case-insensitive)."""
    for entry in sorted(os.listdir(directory)):
        full = os.path.join(directory, entry)
        if os.path.isdir(full) and entry.lower() in wanted_names:
            return full
    return None


def _has_images(directory):
    """True if the folder contains class subfolders that hold image files."""
    for entry in sorted(os.listdir(directory)):
        sub = os.path.join(directory, entry)
        if os.path.isdir(sub):
            for f in os.listdir(sub):
                if f.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff")):
                    return True
    return False


def find_dataset_layout(base_dir, max_depth=3):
    """Return (source_dir, test_dir). test_dir is None if there is no test folder.

    Searches inside base_dir for a folder that contains train/ and test/
    subfolders, so it works no matter how deeply the archive was extracted.
    """
    if not os.path.isdir(base_dir):
        raise FileNotFoundError(f"Dataset folder '{base_dir}' not found.")

    # Walk the tree (shallow first) looking for a train/test pair.
    for root, dirs, _ in os.walk(base_dir):
        depth = root[len(base_dir):].count(os.sep)
        if depth > max_depth:
            dirs[:] = []          # stop descending further
            continue

        train_dir = _match_subdir(root, TRAIN_NAMES)
        test_dir  = _match_subdir(root, TEST_NAMES)
        if train_dir and test_dir:
            return train_dir, test_dir

    # No train/test pair found - look for a folder holding the class subfolders.
    for root, dirs, _ in os.walk(base_dir):
        depth = root[len(base_dir):].count(os.sep)
        if depth > max_depth:
            dirs[:] = []
            continue
        if _has_images(root):
            return root, None

    raise FileNotFoundError(
        f"Could not find image class folders inside '{base_dir}'. "
        f"Check that the dataset was extracted correctly."
    )


SOURCE_DIR, TEST_DIR = find_dataset_layout(DATA_DIR)
print("Source folder:", SOURCE_DIR)
print("Test folder  :", TEST_DIR if TEST_DIR else "(none - test set is split from source)")


class MRIDataset(Dataset):
    """Loads images from a list of (path, label) pairs and applies a transform."""

    def __init__(self, samples, transform):
        self.samples = samples
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path, label = self.samples[idx]
        image = Image.open(path).convert("RGB")
        return self.transform(image), label


# Read all image paths and labels first (no transform applied yet).
base_folder = datasets.ImageFolder(SOURCE_DIR)
CLASS_NAMES = base_folder.classes
NUM_CLASSES = len(CLASS_NAMES)

all_samples = base_folder.samples
all_labels  = [label for _, label in all_samples]

if TEST_DIR is not None:
    # Testing folder already exists -> split source into train (85%) / val (15%)
    train_samples, val_samples = train_test_split(
        all_samples, test_size=0.15, random_state=SEED, stratify=all_labels)
    test_samples = datasets.ImageFolder(TEST_DIR).samples
else:
    # No Testing folder -> make a stratified 70/15/15 split
    train_samples, temp_samples = train_test_split(
        all_samples, test_size=0.30, random_state=SEED, stratify=all_labels)
    temp_labels = [label for _, label in temp_samples]
    val_samples, test_samples = train_test_split(
        temp_samples, test_size=0.50, random_state=SEED, stratify=temp_labels)

# Training data uses augmentation; validation and test do not.
train_set = MRIDataset(train_samples, train_transform)
val_set   = MRIDataset(val_samples,   eval_transform)
test_set  = MRIDataset(test_samples,  eval_transform)

print("Classes:", CLASS_NAMES)
print("Train:", len(train_set), "| Validation:", len(val_set), "| Test:", len(test_set))


# Create data loaders.
train_loader = DataLoader(train_set, batch_size=BATCH_SIZE, shuffle=True,  num_workers=2)
val_loader   = DataLoader(val_set,   batch_size=BATCH_SIZE, shuffle=False, num_workers=2)
test_loader  = DataLoader(test_set,  batch_size=BATCH_SIZE, shuffle=False, num_workers=2)


# ## 3. Exploratory Data Analysis (EDA)


# Class distribution in the training folder.
counts = [0] * NUM_CLASSES
for _, label in all_samples:
    counts[label] += 1

plt.figure(figsize=(6, 4))
plt.bar(CLASS_NAMES, counts, color="steelblue")
plt.title("Class distribution (training folder)")
plt.ylabel("Number of images")
plt.show()

for name, c in zip(CLASS_NAMES, counts):
    print(f"{name:12s}: {c}")


# Show one sample image from each class.
fig, axes = plt.subplots(1, NUM_CLASSES, figsize=(12, 3))

for i, class_name in enumerate(CLASS_NAMES):
    # find the first image belonging to this class
    path = next(p for p, lbl in all_samples if lbl == i)
    img = plt.imread(path)
    axes[i].imshow(img, cmap="gray")
    axes[i].set_title(class_name)
    axes[i].axis("off")

plt.tight_layout()
plt.show()


# ## 4. Model Setup
#
# We use three pretrained models from torchvision. For each one, the original
# classifier is replaced with a new head that outputs 4 classes.


def build_baseline(name):
    """Build a pretrained baseline model with a new 4-class classifier head."""

    if name == "VGG16":
        model = models.vgg16(weights=models.VGG16_Weights.DEFAULT)
        model.classifier[6] = nn.Linear(4096, NUM_CLASSES)

    elif name == "ResNet50":
        model = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        model.fc = nn.Linear(model.fc.in_features, NUM_CLASSES)

    elif name == "DenseNet121":
        model = models.densenet121(weights=models.DenseNet121_Weights.DEFAULT)
        model.classifier = nn.Linear(model.classifier.in_features, NUM_CLASSES)

    else:
        raise ValueError("Unknown model name: " + name)

    return model.to(DEVICE)


# ## 5. Training and Evaluation Functions


def train_one_epoch(model, loader, criterion, optimizer):
    """Run one training epoch and return the average loss and accuracy."""
    model.train()
    total_loss, correct, total = 0.0, 0, 0

    for images, labels in loader:
        images, labels = images.to(DEVICE), labels.to(DEVICE)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * images.size(0)
        correct    += (outputs.argmax(1) == labels).sum().item()
        total      += labels.size(0)

    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, criterion):
    """Evaluate the model and return loss, accuracy, predictions, labels and probabilities."""
    model.eval()
    total_loss, correct, total = 0.0, 0, 0
    all_preds, all_labels, all_probs = [], [], []

    for images, labels in loader:
        images, labels = images.to(DEVICE), labels.to(DEVICE)

        outputs = model(images)
        loss = criterion(outputs, labels)
        probs = torch.softmax(outputs, dim=1)

        total_loss += loss.item() * images.size(0)
        correct    += (outputs.argmax(1) == labels).sum().item()
        total      += labels.size(0)

        all_preds.extend(outputs.argmax(1).cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
        all_probs.extend(probs.cpu().numpy())

    return (total_loss / total, correct / total,
            np.array(all_preds), np.array(all_labels), np.array(all_probs))


def train_model(name, epochs=EPOCHS, lr=LR):
    """Train one baseline model and return the trained model and its training history."""
    print(f"\n===== Training {name} =====")

    set_seed(SEED)
    model = build_baseline(name)

    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)

    history = {"train_loss": [], "train_acc": [], "val_loss": [], "val_acc": []}

    for epoch in range(1, epochs + 1):
        tr_loss, tr_acc = train_one_epoch(model, train_loader, criterion, optimizer)
        va_loss, va_acc, _, _, _ = evaluate(model, val_loader, criterion)

        history["train_loss"].append(tr_loss)
        history["train_acc"].append(tr_acc)
        history["val_loss"].append(va_loss)
        history["val_acc"].append(va_acc)

        print(f"Epoch {epoch:2d}/{epochs} | "
              f"train loss {tr_loss:.4f} acc {tr_acc:.4f} | "
              f"val loss {va_loss:.4f} acc {va_acc:.4f}")

    return model, history


def compute_metrics(y_true, y_pred, y_prob):
    """Compute all evaluation metrics for the test set."""
    return {
        "accuracy":         float(accuracy_score(y_true, y_pred)),
        "precision_macro":  float(precision_score(y_true, y_pred, average="macro", zero_division=0)),
        "recall_macro":     float(recall_score(y_true, y_pred, average="macro", zero_division=0)),
        "f1_macro":         float(f1_score(y_true, y_pred, average="macro", zero_division=0)),
        "roc_auc_ovr":      float(roc_auc_score(y_true, y_prob, multi_class="ovr")),
        "confusion_matrix": confusion_matrix(y_true, y_pred).tolist(),
        "per_class_f1":     f1_score(y_true, y_pred, average=None, zero_division=0).tolist(),
    }


# ## 6. Train the Three Baseline Models


criterion = nn.CrossEntropyLoss()
results = {}

for name in ["VGG16", "ResNet50", "DenseNet121"]:
    model, history = train_model(name)

    # Evaluate on the test set
    test_loss, test_acc, preds, labels, probs = evaluate(model, test_loader, criterion)
    metrics = compute_metrics(labels, preds, probs)

    results[name] = {
        "history": history,
        "metrics": metrics,
        "test_predictions": preds.tolist(),   # needed for McNemar's test in Part 2
        "test_labels": labels.tolist(),
    }

    print(f"\n{name} test accuracy: {metrics['accuracy']:.4f} | "
          f"F1 (macro): {metrics['f1_macro']:.4f}")


# ## 7. Results and Visualisation


# Summary table of all baseline results.
print(f"{'Model':<14}{'Accuracy':>10}{'Precision':>11}{'Recall':>9}{'F1':>9}{'ROC-AUC':>10}")
print("-" * 63)

for name, r in results.items():
    m = r["metrics"]
    print(f"{name:<14}{m['accuracy']:>10.4f}{m['precision_macro']:>11.4f}"
          f"{m['recall_macro']:>9.4f}{m['f1_macro']:>9.4f}{m['roc_auc_ovr']:>10.4f}")


# Plot training and validation curves for all baselines.
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

for name, r in results.items():
    axes[0].plot(r["history"]["val_loss"], label=name)
    axes[1].plot(r["history"]["val_acc"],  label=name)

axes[0].set_title("Validation loss")
axes[0].set_xlabel("Epoch"); axes[0].set_ylabel("Loss"); axes[0].legend(); axes[0].grid(alpha=0.3)

axes[1].set_title("Validation accuracy")
axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("Accuracy"); axes[1].legend(); axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()


# Plot confusion matrices.
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (name, r) in zip(axes, results.items()):
    cm = np.array(r["metrics"]["confusion_matrix"])
    ax.imshow(cm, cmap="Blues")
    ax.set_title(name)
    ax.set_xticks(range(NUM_CLASSES)); ax.set_xticklabels(CLASS_NAMES, rotation=45, ha="right")
    ax.set_yticks(range(NUM_CLASSES)); ax.set_yticklabels(CLASS_NAMES)
    ax.set_xlabel("Predicted"); ax.set_ylabel("True")

    for i in range(NUM_CLASSES):
        for j in range(NUM_CLASSES):
            ax.text(j, i, cm[i, j], ha="center", va="center",
                    color="white" if cm[i, j] > cm.max() / 2 else "black")

plt.tight_layout()
plt.show()


# Detailed per-class report for each baseline.
for name, r in results.items():
    print(f"\n===== {name} =====")
    print(classification_report(r["test_labels"], r["test_predictions"],
                                target_names=CLASS_NAMES, digits=4,
                                zero_division=0))


# ## 8. Save Results to JSON


# Save everything needed by Part 2 (metrics, curves, predictions).
output = {
    "class_names": CLASS_NAMES,
    "config": {
        "img_size": IMG_SIZE,
        "batch_size": BATCH_SIZE,
        "epochs": EPOCHS,
        "learning_rate": LR,
        "seed": SEED,
    },
    "models": results,
}

out_path = os.path.join(RESULTS_DIR, "baseline_results.json")
with open(out_path, "w") as f:
    json.dump(output, f, indent=2)

print("Saved results to:", out_path)
```

```python
# # Brain Tumor MRI Classification - Part 2: Proposed Model
#
# This notebook covers:
# 1. The proposed CNN model, built and trained completely from scratch (no pretrained weights)
# 2. Training with techniques that keep the training curve stable
# 3. Evaluation and comparison against the baseline models from Part 1
# 4. McNemar's statistical test
#
# **Important:** Run Part 1 first, so that `results/baseline_results.json` exists.
#
# **Proposed architecture:**
# Input -> Conv block 1 -> Conv block 2 -> Residual add -> Conv block 3 ->
# SE attention -> Global average pooling -> Dense classifier -> Softmax (4 classes)


# ## 1. Imports and Configuration


import os
import json
import random
import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset
from torchvision import datasets, transforms
from PIL import Image

from sklearn.model_selection import train_test_split
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, confusion_matrix, classification_report,
                             roc_auc_score)
from statsmodels.stats.contingency_tables import mcnemar

print("PyTorch version:", torch.__version__)
print("GPU available:", torch.cuda.is_available())


# ---------------- Configuration ----------------
# These values match Part 1 so the comparison is fair.

DATA_DIR    = "/content/dataset/dataset/classification_task"
RESULTS_DIR = "/content/results"

IMG_SIZE   = 224
BATCH_SIZE = 32
EPOCHS     = 60          # training from scratch needs many more epochs than fine-tuning
LR         = 1e-3        # AdamW works well with this starting learning rate
WEIGHT_DECAY  = 5e-4
LABEL_SMOOTH  = 0.05     # lowered: the model was underfitting, not overfitting
GRAD_CLIP     = 1.0
WARMUP_EPOCHS = 5
SEED       = 42          # same seed as Part 1 -> same data split

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
os.makedirs(RESULTS_DIR, exist_ok=True)

print("Using device:", DEVICE)


def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)


# ## 2. Data Loading
#
# The same preprocessing, split and seed as Part 1 are used, so the proposed model
# and the baselines are evaluated under identical conditions.


MEAN = [0.485, 0.456, 0.406]
STD  = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])

eval_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])


# ---------------- Load the dataset ----------------
# Some copies of this dataset have Training/Testing folders, others have the four
# class folders directly inside `dataset`. Both layouts are handled here.

TRAIN_NAMES = ["train", "training"]
TEST_NAMES  = ["test", "testing"]


def _match_subdir(directory, wanted_names):
    """Find a subfolder whose name matches one of wanted_names (case-insensitive)."""
    for entry in sorted(os.listdir(directory)):
        full = os.path.join(directory, entry)
        if os.path.isdir(full) and entry.lower() in wanted_names:
            return full
    return None


def _has_images(directory):
    """True if the folder contains class subfolders that hold image files."""
    for entry in sorted(os.listdir(directory)):
        sub = os.path.join(directory, entry)
        if os.path.isdir(sub):
            for f in os.listdir(sub):
                if f.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff")):
                    return True
    return False


def find_dataset_layout(base_dir, max_depth=3):
    """Return (source_dir, test_dir). test_dir is None if there is no test folder.

    Searches inside base_dir for a folder that contains train/ and test/
    subfolders, so it works no matter how deeply the archive was extracted.
    """
    if not os.path.isdir(base_dir):
        raise FileNotFoundError(f"Dataset folder '{base_dir}' not found.")

    # Walk the tree (shallow first) looking for a train/test pair.
    for root, dirs, _ in os.walk(base_dir):
        depth = root[len(base_dir):].count(os.sep)
        if depth > max_depth:
            dirs[:] = []          # stop descending further
            continue

        train_dir = _match_subdir(root, TRAIN_NAMES)
        test_dir  = _match_subdir(root, TEST_NAMES)
        if train_dir and test_dir:
            return train_dir, test_dir

    # No train/test pair found - look for a folder holding the class subfolders.
    for root, dirs, _ in os.walk(base_dir):
        depth = root[len(base_dir):].count(os.sep)
        if depth > max_depth:
            dirs[:] = []
            continue
        if _has_images(root):
            return root, None

    raise FileNotFoundError(
        f"Could not find image class folders inside '{base_dir}'. "
        f"Check that the dataset was extracted correctly."
    )


SOURCE_DIR, TEST_DIR = find_dataset_layout(DATA_DIR)
print("Source folder:", SOURCE_DIR)
print("Test folder  :", TEST_DIR if TEST_DIR else "(none - test set is split from source)")


class MRIDataset(Dataset):
    """Loads images from a list of (path, label) pairs and applies a transform."""

    def __init__(self, samples, transform):
        self.samples = samples
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path, label = self.samples[idx]
        image = Image.open(path).convert("RGB")
        return self.transform(image), label


# Read all image paths and labels first (no transform applied yet).
base_folder = datasets.ImageFolder(SOURCE_DIR)
CLASS_NAMES = base_folder.classes
NUM_CLASSES = len(CLASS_NAMES)

all_samples = base_folder.samples
all_labels  = [label for _, label in all_samples]

if TEST_DIR is not None:
    # Testing folder already exists -> split source into train (85%) / val (15%)
    train_samples, val_samples = train_test_split(
        all_samples, test_size=0.15, random_state=SEED, stratify=all_labels)
    test_samples = datasets.ImageFolder(TEST_DIR).samples
else:
    # No Testing folder -> make a stratified 70/15/15 split
    train_samples, temp_samples = train_test_split(
        all_samples, test_size=0.30, random_state=SEED, stratify=all_labels)
    temp_labels = [label for _, label in temp_samples]
    val_samples, test_samples = train_test_split(
        temp_samples, test_size=0.50, random_state=SEED, stratify=temp_labels)

# Training data uses augmentation; validation and test do not.
train_set = MRIDataset(train_samples, train_transform)
val_set   = MRIDataset(val_samples,   eval_transform)
test_set  = MRIDataset(test_samples,  eval_transform)

print("Classes:", CLASS_NAMES)
print("Train:", len(train_set), "| Validation:", len(val_set), "| Test:", len(test_set))

train_loader = DataLoader(train_set, batch_size=BATCH_SIZE, shuffle=True,  num_workers=2)
val_loader   = DataLoader(val_set,   batch_size=BATCH_SIZE, shuffle=False, num_workers=2)
test_loader  = DataLoader(test_set,  batch_size=BATCH_SIZE, shuffle=False, num_workers=2)




# ## 3. Proposed Model Architecture
#
# The model has three building parts:
#
# - **SEBlock**: channel attention. It learns how important each feature channel is,
#   so the model focuses more on tumor regions and less on background.
# - **ResidualSEBlock**: two 3x3 convolutions + SE attention inside a residual
#   connection. The skip connection keeps gradients flowing, which is what makes a
#   deep network trainable from scratch.
# - **ProposedCNN**: a stem followed by four stages of residual SE blocks, with the
#   channel count doubling and the resolution halving at each stage.
#
# Distinguishing glioma from meningioma needs deep, high-level features, so the
# network is built with enough depth and width to actually learn them.


class SEBlock(nn.Module):
    """Squeeze-and-Excitation attention: learns an importance weight per channel."""

    def __init__(self, channels, reduction=16):
        super().__init__()
        hidden = max(channels // reduction, 8)
        self.fc1 = nn.Linear(channels, hidden)
        self.fc2 = nn.Linear(hidden, channels)

    def forward(self, x):
        b, c, _, _ = x.shape
        # Squeeze: average each channel map into a single number
        w = F.adaptive_avg_pool2d(x, 1).view(b, c)
        # Excitation: learn how important each channel is
        w = F.relu(self.fc1(w))
        w = torch.sigmoid(self.fc2(w)).view(b, c, 1, 1)
        # Rescale the original feature maps
        return x * w


class ResidualSEBlock(nn.Module):
    """Two 3x3 convolutions + SE attention, wrapped in a residual connection.

    The residual (skip) connection lets gradients flow directly through the
    network, which is what makes a deep from-scratch model trainable.
    """

    def __init__(self, in_channels, out_channels, stride=1):
        super().__init__()

        self.conv1 = nn.Conv2d(in_channels, out_channels, 3, stride=stride, padding=1, bias=False)
        self.bn1   = nn.BatchNorm2d(out_channels)
        self.conv2 = nn.Conv2d(out_channels, out_channels, 3, padding=1, bias=False)
        self.bn2   = nn.BatchNorm2d(out_channels)
        self.se    = SEBlock(out_channels)

        # If the shape changes, the shortcut needs a 1x1 convolution to match it.
        if stride != 1 or in_channels != out_channels:
            self.shortcut = nn.Sequential(
                nn.Conv2d(in_channels, out_channels, 1, stride=stride, bias=False),
                nn.BatchNorm2d(out_channels),
            )
        else:
            self.shortcut = nn.Identity()

    def forward(self, x):
        out = F.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        out = self.se(out)                 # channel attention
        return F.relu(out + self.shortcut(x))   # residual add


class ProposedCNN(nn.Module):
    """Custom CNN trained from scratch: a deep stack of residual + SE blocks.

    Layout (for a 224x224 input):
        stem     3   -> 64    224 -> 56
        stage 1  64  -> 64     56 -> 56
        stage 2  64  -> 128    56 -> 28
        stage 3  128 -> 256    28 -> 14
        stage 4  256 -> 512    14 -> 7
        global average pooling -> dropout -> linear -> 4 classes
    """

    def __init__(self, num_classes=4, width=64, blocks_per_stage=2):
        super().__init__()

        self.stem = nn.Sequential(
            nn.Conv2d(3, width, 3, stride=2, padding=1, bias=False),
            nn.BatchNorm2d(width),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),
        )

        self.stage1 = self._make_stage(width,     width,     blocks_per_stage, stride=1)
        self.stage2 = self._make_stage(width,     width * 2, blocks_per_stage, stride=2)
        self.stage3 = self._make_stage(width * 2, width * 4, blocks_per_stage, stride=2)
        self.stage4 = self._make_stage(width * 4, width * 8, blocks_per_stage, stride=2)

        self.head = nn.Sequential(
            nn.Dropout(0.3),
            nn.Linear(width * 8, num_classes),
        )

        self._init_weights()

    def _make_stage(self, in_channels, out_channels, n_blocks, stride):
        """Build one stage: the first block may downsample, the rest keep the size."""
        layers = [ResidualSEBlock(in_channels, out_channels, stride)]
        for _ in range(n_blocks - 1):
            layers.append(ResidualSEBlock(out_channels, out_channels, 1))
        return nn.Sequential(*layers)

    def _init_weights(self):
        # He initialisation works well with ReLU and helps training start stably.
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                nn.init.kaiming_normal_(m.weight, mode="fan_out", nonlinearity="relu")
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.constant_(m.weight, 1)
                nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.Linear):
                nn.init.normal_(m.weight, 0, 0.01)
                nn.init.constant_(m.bias, 0)

    def forward(self, x):
        x = self.stem(x)
        x = self.stage1(x)
        x = self.stage2(x)
        x = self.stage3(x)
        x = self.stage4(x)
        x = F.adaptive_avg_pool2d(x, 1).flatten(1)   # global average pooling
        return self.head(x)


# Quick check: the model should output a tensor of shape (batch, 4).
model = ProposedCNN(NUM_CLASSES).to(DEVICE)
dummy = torch.randn(2, 3, IMG_SIZE, IMG_SIZE).to(DEVICE)
print("Output shape:", model(dummy).shape)
print("Trainable parameters:", sum(p.numel() for p in model.parameters() if p.requires_grad))


# ## 4. Training Setup
#
# To keep the training curve smooth instead of fluctuating, the following are used:
#
# | Technique | Purpose |
# |---|---|
# | AdamW optimiser | stable updates with weight decay |
# | Warmup + cosine annealing | avoids large jumps early, smooth decay later |
# | Gradient clipping | prevents sudden spikes in the loss |
# | Label smoothing | softer targets, less overconfident jumps |
# | Batch normalisation | already in every conv block |
# | He initialisation | stable starting point |


def train_one_epoch(model, loader, criterion, optimizer):
    model.train()
    total_loss, correct, total = 0.0, 0, 0

    for images, labels in loader:
        images, labels = images.to(DEVICE), labels.to(DEVICE)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()

        # Gradient clipping keeps the updates from becoming too large.
        torch.nn.utils.clip_grad_norm_(model.parameters(), GRAD_CLIP)
        optimizer.step()

        total_loss += loss.item() * images.size(0)
        correct    += (outputs.argmax(1) == labels).sum().item()
        total      += labels.size(0)

    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, criterion):
    model.eval()
    total_loss, correct, total = 0.0, 0, 0
    all_preds, all_labels, all_probs = [], [], []

    for images, labels in loader:
        images, labels = images.to(DEVICE), labels.to(DEVICE)

        outputs = model(images)
        loss = criterion(outputs, labels)
        probs = torch.softmax(outputs, dim=1)

        total_loss += loss.item() * images.size(0)
        correct    += (outputs.argmax(1) == labels).sum().item()
        total      += labels.size(0)

        all_preds.extend(outputs.argmax(1).cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
        all_probs.extend(probs.cpu().numpy())

    return (total_loss / total, correct / total,
            np.array(all_preds), np.array(all_labels), np.array(all_probs))


def build_scheduler(optimizer, epochs, warmup_epochs):
    """Learning rate: linear warmup for the first few epochs, then cosine decay."""

    def lr_lambda(epoch):
        if epoch < warmup_epochs:
            return (epoch + 1) / warmup_epochs
        progress = (epoch - warmup_epochs) / max(1, epochs - warmup_epochs)
        return 0.5 * (1 + np.cos(np.pi * progress))

    return torch.optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)


# ## 5. Train the Proposed Model


set_seed(SEED)
model = ProposedCNN(NUM_CLASSES).to(DEVICE)

criterion = nn.CrossEntropyLoss(label_smoothing=LABEL_SMOOTH)
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
scheduler = build_scheduler(optimizer, EPOCHS, WARMUP_EPOCHS)

history = {"train_loss": [], "train_acc": [], "val_loss": [], "val_acc": [], "lr": []}
best_val_acc, best_state = 0.0, None

for epoch in range(1, EPOCHS + 1):
    tr_loss, tr_acc = train_one_epoch(model, train_loader, criterion, optimizer)
    va_loss, va_acc, _, _, _ = evaluate(model, val_loader, criterion)

    current_lr = optimizer.param_groups[0]["lr"]
    scheduler.step()

    history["train_loss"].append(tr_loss)
    history["train_acc"].append(tr_acc)
    history["val_loss"].append(va_loss)
    history["val_acc"].append(va_acc)
    history["lr"].append(current_lr)

    # Keep the weights from the best epoch.
    if va_acc > best_val_acc:
        best_val_acc = va_acc
        best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}

    print(f"Epoch {epoch:2d}/{EPOCHS} | lr {current_lr:.6f} | "
          f"train loss {tr_loss:.4f} acc {tr_acc:.4f} | "
          f"val loss {va_loss:.4f} acc {va_acc:.4f}")

print(f"\nBest validation accuracy: {best_val_acc:.4f}")


# Load the best weights before final evaluation.
model.load_state_dict(best_state)
model.to(DEVICE)
print("Loaded best model weights.")


# ## 6. Evaluation on the Test Set


def compute_metrics(y_true, y_pred, y_prob):
    """Compute all evaluation metrics for the test set."""
    return {
        "accuracy":         float(accuracy_score(y_true, y_pred)),
        "precision_macro":  float(precision_score(y_true, y_pred, average="macro", zero_division=0)),
        "recall_macro":     float(recall_score(y_true, y_pred, average="macro", zero_division=0)),
        "f1_macro":         float(f1_score(y_true, y_pred, average="macro", zero_division=0)),
        "roc_auc_ovr":      float(roc_auc_score(y_true, y_prob, multi_class="ovr")),
        "confusion_matrix": confusion_matrix(y_true, y_pred).tolist(),
        "per_class_f1":     f1_score(y_true, y_pred, average=None, zero_division=0).tolist(),
    }


# Evaluate the trained model on the test set.
test_loss, test_acc, preds, labels, probs = evaluate(model, test_loader, criterion)
proposed_metrics = compute_metrics(labels, preds, probs)

print("Proposed model test results")
print("---------------------------")
for key in ["accuracy", "precision_macro", "recall_macro", "f1_macro", "roc_auc_ovr"]:
    print(f"{key:18s}: {proposed_metrics[key]:.4f}")


print(classification_report(labels, preds, target_names=CLASS_NAMES, digits=4,
                                zero_division=0))


# ## 7. Training Curves


fig, axes = plt.subplots(1, 3, figsize=(16, 4))

axes[0].plot(history["train_loss"], label="Train")
axes[0].plot(history["val_loss"],   label="Validation")
axes[0].set_title("Loss"); axes[0].set_xlabel("Epoch"); axes[0].legend(); axes[0].grid(alpha=0.3)

axes[1].plot(history["train_acc"], label="Train")
axes[1].plot(history["val_acc"],   label="Validation")
axes[1].set_title("Accuracy"); axes[1].set_xlabel("Epoch"); axes[1].legend(); axes[1].grid(alpha=0.3)

axes[2].plot(history["lr"], color="darkorange")
axes[2].set_title("Learning rate schedule"); axes[2].set_xlabel("Epoch"); axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()


# Confusion matrix of the proposed model.
cm = np.array(proposed_metrics["confusion_matrix"])

plt.figure(figsize=(5.5, 5))
plt.imshow(cm, cmap="Blues")
plt.title("Proposed model - confusion matrix")
plt.xticks(range(NUM_CLASSES), CLASS_NAMES, rotation=45, ha="right")
plt.yticks(range(NUM_CLASSES), CLASS_NAMES)
plt.xlabel("Predicted"); plt.ylabel("True")

for i in range(NUM_CLASSES):
    for j in range(NUM_CLASSES):
        plt.text(j, i, cm[i, j], ha="center", va="center",
                 color="white" if cm[i, j] > cm.max() / 2 else "black")

plt.tight_layout()
plt.show()


# ## 8. Comparison with the Baseline Models


# Load the baseline results saved by Part 1.
with open(os.path.join(RESULTS_DIR, "baseline_results.json")) as f:
    baseline_data = json.load(f)

baselines = baseline_data["models"]
print("Loaded baselines:", list(baselines.keys()))


# Comparison table.
print(f"{'Model':<16}{'Accuracy':>10}{'Precision':>11}{'Recall':>9}{'F1':>9}{'ROC-AUC':>10}")
print("-" * 65)

for name, r in baselines.items():
    m = r["metrics"]
    print(f"{name:<16}{m['accuracy']:>10.4f}{m['precision_macro']:>11.4f}"
          f"{m['recall_macro']:>9.4f}{m['f1_macro']:>9.4f}{m['roc_auc_ovr']:>10.4f}")

m = proposed_metrics
print(f"{'Proposed CNN':<16}{m['accuracy']:>10.4f}{m['precision_macro']:>11.4f}"
      f"{m['recall_macro']:>9.4f}{m['f1_macro']:>9.4f}{m['roc_auc_ovr']:>10.4f}")


# Bar chart comparing accuracy and F1-score.
names  = list(baselines.keys()) + ["Proposed CNN"]
accs   = [baselines[n]["metrics"]["accuracy"] for n in baselines] + [proposed_metrics["accuracy"]]
f1s    = [baselines[n]["metrics"]["f1_macro"] for n in baselines] + [proposed_metrics["f1_macro"]]

x = np.arange(len(names))
width = 0.35

plt.figure(figsize=(9, 4.5))
plt.bar(x - width/2, accs, width, label="Accuracy")
plt.bar(x + width/2, f1s,  width, label="F1 (macro)")
plt.xticks(x, names, rotation=15)
plt.ylim(0.8, 1.0)
plt.ylabel("Score")
plt.title("Proposed model vs baselines")
plt.legend(); plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()


# ## 9. McNemar's Statistical Test
#
# McNemar's test checks whether the difference between two models on the same test set
# is statistically significant, or could just be random chance.
#
# - Null hypothesis: both models have the same error rate.
# - If p-value < 0.05, the difference is statistically significant.


def mcnemar_test(preds_a, preds_b, true_labels):
    """Compare two models on the same test set using McNemar's test.

    The contingency table counts how the two models agree or disagree:
        only_a = model A correct, model B wrong
        only_b = model A wrong,   model B correct
    Only these two cells decide the result.
    """
    correct_a = (np.array(preds_a) == np.array(true_labels))
    correct_b = (np.array(preds_b) == np.array(true_labels))

    both_correct = int(np.sum(correct_a & correct_b))
    only_a       = int(np.sum(correct_a & ~correct_b))
    only_b       = int(np.sum(~correct_a & correct_b))
    both_wrong   = int(np.sum(~correct_a & ~correct_b))

    table = [[both_correct, only_a],
             [only_b,       both_wrong]]

    # If the two models never disagree, the test is undefined.
    # In that case there is no difference to detect, so p = 1.0.
    if only_a + only_b == 0:
        return 0.0, 1.0, table

    # Use the exact binomial test when disagreements are few (more reliable),
    # otherwise the chi-square version with continuity correction.
    use_exact = (only_a + only_b) < 25
    result = mcnemar(table, exact=use_exact, correction=not use_exact)

    return float(result.statistic), float(result.pvalue), table


# Compare the proposed model against each baseline.
test_labels = np.array(labels)
mcnemar_results = {}

print(f"{'Comparison':<34}{'Statistic':>12}{'p-value':>12}  Significant")
print("-" * 74)

for name, r in baselines.items():
    stat, p, table = mcnemar_test(preds, r["test_predictions"], test_labels)
    significant = p < 0.05

    mcnemar_results[f"Proposed_vs_{name}"] = {
        "statistic": stat,
        "p_value": p,
        "significant_at_0.05": bool(significant),
        "contingency_table": [[int(v) for v in row] for row in table],
    }

    print(f"{'Proposed vs ' + name:<34}{stat:>12.4f}{p:>12.6f}  {'Yes' if significant else 'No'}")


# ## 10. Save Results to JSON


output = {
    "class_names": CLASS_NAMES,
    "config": {
        "img_size": IMG_SIZE,
        "batch_size": BATCH_SIZE,
        "epochs": EPOCHS,
        "learning_rate": LR,
        "weight_decay": WEIGHT_DECAY,
        "label_smoothing": LABEL_SMOOTH,
        "grad_clip": GRAD_CLIP,
        "warmup_epochs": WARMUP_EPOCHS,
        "seed": SEED,
    },
    "proposed_model": {
        "history": history,
        "metrics": proposed_metrics,
        "test_predictions": preds.tolist(),
        "test_labels": test_labels.tolist(),
        "best_val_accuracy": float(best_val_acc),
        "num_parameters": int(sum(p.numel() for p in model.parameters())),
    },
    "mcnemar_test": mcnemar_results,
}

out_path = os.path.join(RESULTS_DIR, "proposed_model_results.json")
with open(out_path, "w") as f:
    json.dump(output, f, indent=2)

print("Saved results to:", out_path)


# ## 11. Summary
#
# The proposed CNN was trained entirely from scratch, without any pretrained weights,
# and compared against three pretrained baselines under identical conditions
# (same split, same preprocessing, same augmentation, same seed).
#
# The stability techniques (warmup + cosine annealing, gradient clipping, label
# smoothing, batch normalisation, He initialisation) keep the training curve smooth
# instead of fluctuating. McNemar's test in section 9 shows whether the performance
# difference against each baseline is statistically significant.
#
# Note on fairness: the proposed model is trained for more epochs than the
# fine-tuned baselines. This is expected and should be stated in the report -
# a randomly initialised network needs far more epochs to converge than one that
# starts from ImageNet weights. Everything else (data split, preprocessing,
# augmentation, evaluation) is identical.
```
